```python
#student performance analysis
import numpy as np
import pandas as pd

student=['studentID','Name','Department','Subject','marks','attendance']
data=[103,'Ashutosh','CSE','DBMS',90,75],[104,'RAM','CSE','DBMS',85,70],[105,'VIJAY','CSE','DBMS',80,78],[106,'RAVI','CSE','DBMS',75,72],[107,'RAJU','CSE','DBMS',65,60],[108,'RAMESH','CSE','OS',95,85],[109,'TILAK','CSE','OS',90,80],[110,'ATHARVA','CSE','OS',85,75],[111,'SANJU','CSE','OS',80,70],[112,'VIRAT','CSE','OS',75,65],[113,'DAVID','CSE','OS',60,60],[114,'SACHIN','CSE','CIVIL',85,80],[115,'ARSHDEEP','CSE','CIVIL',80,75],[116,'SHAMI','CSE','CIVIL',75,70],[117,'SIDHU','CSE','CIVIL',70,65],[118,'SEHWAG','CSE','CIVIL',65,60]
print(data)
df=pd.DataFrame(data, columns=student)
data
df.head()
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)
print(df.isnull().sum())

# Showing name and marks directly
df[["Name", "marks"]]

#calculating average marks
df["Average marks"] = df["marks"].mean()
df[["Name", "marks", "Average marks"]]

def assign_grade(marks):
  if marks>=90:
    return "A+"
  elif marks>=85:
    return "A"
  elif marks>=80:
    return "B"
  elif marks>=75:
    return "C"
  elif marks>=70:
    return "D"
  else:
    return "F"

# Applying the grading function to the marks column
df["Grade"] = df["marks"].apply(assign_grade)
display(df[["Name", "marks", "Grade"]])

def check_result(row):
  # Check if the student's marks in their subject are less than 40
  if row["marks"] < 40:
        return "Fail"

  # Check if attendance is less than 75
  if row["attendance"] < 75:
        return "Fail"

  return "Pass"

df["Result"] = df.apply(check_result, axis=1)

display(df[["Name", "attendance", "Average marks", "Result"]])

#highest performing student based on individual marks
highest_marks = df["marks"].max()

top_student = df.loc[
    df["marks"].idxmax()
]

print("Highest Marks:", highest_marks)
print("Top Student:", top_student["Name"])

lowest_student = df.loc[
    df["marks"].idxmin()
]

print("Lowest Performing Student:", lowest_student["Name"])
print("Lowest Marks:", lowest_student["marks"])
print("Average Marks:", lowest_student["Average marks"])

class_average = np.mean(df["Average marks"])

print("Class Average:",
      round(class_average, 2))

subjects = df["Subject"].unique()
for subject in subjects:
    average = df[df["Subject"] == subject]["marks"].mean()
    print(
        subject,
        "Average =",
        round(average, 2)
    )

for subject in subjects:
    highest = df[df["Subject"] == subject]["marks"].max()
    lowest = df[df["Subject"] == subject]["marks"].min()
    print("\n" + subject)
    print("Highest:", highest)
    print("Lowest:", lowest)

top_students = df.sort_values(
    by="marks",
    ascending=False
).head(3)

display(top_students[
    ["Name", "marks", "Grade"]
])

pass_count = (df["Result"] == "Pass").sum()
fail_count = (df["Result"] == "Fail").sum()

print("Pass Students:", pass_count)
print("Fail Students:", fail_count)

final_report = df[
    [
        "studentID",
        "Name",
        "Subject",
        "attendance",
        "marks",
        "Average marks",
        "Grade",
        "Result"
    ]
]

display(final_report)
```
